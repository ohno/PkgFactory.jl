# Create a Julia package with PkgFactory

This notebook previews the package before making any GitHub changes. Authentication uses GitHub's OAuth device flow, and the access token is never displayed.

```julia
import Pkg
Pkg.activate(".")
Pkg.add(url="https://github.com/ohno/PkgFactory.jl.git")
```

```julia
import PkgFactory
```

```julia
config = PkgFactory.PackageConfig(
    owner = "octocat",
    name = "MyPkg",
    authors = ["The Octocat"],
    description = "A package created from Jupyter",
)
```

```julia
plan = PkgFactory.preview(config)
display(plan)
```

Open the URL printed by the next cell, enter the one-time code, and authorize PkgFactory. The cell waits until authorization completes.

```julia
github = PkgFactory.github_device_login()
```

```julia
result = PkgFactory.create!(plan; backend = github)
```
